```python
import pandas as pd

# Each makeup product is an object (OOP).
class MakeupProduct:
    def __init__(self, name, category, skin_types, looks, price, ingredients):
        self.name = name
        self.category = category
        self.skin_types = skin_types
        self.looks = looks
        self.price = price
        self.ingredients = ingredients

# Practice products: not real brands or verified ingredient lists.
products = [
    MakeupProduct("Hydrating Base", "Base", ["dry", "normal", "combination"], ["natural", "glam"], 18, ["glycerin"]),
    MakeupProduct("Matte Base", "Base", ["oily", "combination"], ["natural", "glam"], 16, ["silica"]),
    MakeupProduct("Cream Blush", "Blush", ["dry", "normal", "combination"], ["natural", "glam"], 12, ["glycerin"]),
    MakeupProduct("Powder Blush", "Blush", ["oily", "combination"], ["natural", "glam"], 10, ["talc"]),
    MakeupProduct("Everyday Mascara", "Mascara", ["dry", "oily", "normal", "combination"], ["natural", "glam"], 11, ["wax"]),
    MakeupProduct("Bold Mascara", "Mascara", ["dry", "oily", "normal", "combination"], ["glam"], 15, ["wax"]),
    MakeupProduct("Lip Gloss", "Lips", ["dry", "oily", "normal", "combination"], ["natural"], 9, ["oil"]),
    MakeupProduct("Lipstick", "Lips", ["dry", "oily", "normal", "combination"], ["glam"], 14, ["wax"])
]

def recommend_products(skin, look, avoid):
    routine = []
    categories = ["Base", "Blush", "Mascara", "Lips"]

    for category in categories:
        best_product = None

        for product in products:
            # Check whether the product matches the user's preferences.
            if product.category == category and skin in product.skin_types and look in product.looks:
                safe_match = True

                # Skip products with an ingredient the user wants to avoid.
                for ingredient in avoid:
                    if ingredient in product.ingredients:
                        safe_match = False

                if safe_match:
                    if best_product == None:
                        best_product = product
                    elif product.price < best_product.price:
                        best_product = product

        if best_product != None:
            routine.append(best_product)

    return routine

def main():
    print("Welcome to Makeup Match!")
    skin = input("Skin type (dry/oily/normal/combination): ").lower().strip()
    look = input("Makeup look (natural/glam): ").lower().strip()

    try:
        budget = float(input("Your total budget ($): "))
    except ValueError:
        print("Please enter a number for your budget.")
        return

    if skin not in ["dry", "oily", "normal", "combination"]:
        print("Please enter a valid skin type.")
        return
    if look not in ["natural", "glam"]:
        print("Please enter a valid makeup look.")
        return
    if budget < 0:
        print("Budget cannot be negative.")
        return

    # For now, the user can enter one ingredient to avoid.
    ingredient = input("One ingredient to avoid (or press Enter for none): ").lower().strip()
    avoid = []
    if ingredient != "":
        avoid.append(ingredient)

    routine = recommend_products(skin, look, avoid)

    if len(routine) < 4:
        print("We could not find a complete routine for your preferences.")
        return

    names = []
    categories = []
    prices = []
    total = 0

    for product in routine:
        names.append(product.name)
        categories.append(product.category)
        prices.append(product.price)
        total = total + product.price

    if total > budget:
        print("The matching routine costs $", total)
        print("This is above your budget of $", budget)
        return

    table = pd.DataFrame({"Step": categories, "Product": names, "Price ($)": prices})
    print("\nYour personalized makeup routine:")
    print(table.to_string(index=False))
    print("Total cost: $", total)
    print("Note: These are fictional practice products, not verified allergy-safe recommendations.")

main()
```
